- 主要数值：保留原摘要关键数字（$1.475M、12.4% 票价收入、市场份额 15.22%→12.19%、Great Lakes 68.4% 等）**须与正文一致**。  
- Keywords: dynamic profit; multidimensional knapsack; gravity model; dynamic pricing; injury state; WNBA; cross-league institutions; behavioral economics; machine learning (optional)
